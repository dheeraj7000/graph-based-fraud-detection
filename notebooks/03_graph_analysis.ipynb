# Local sampled graph analysis

Generate the deterministic local graph with `python scripts/run_sampled_graph_features.py`. This notebook explores the sampled training account features; no database service is required.

```python
import pandas as pd
import matplotlib.pyplot as plt

accounts = pd.read_parquet('data/features/sampled_graph_account_features.parquet')
accounts.describe(include='all')
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
accounts['in_degree'].clip(upper=accounts['in_degree'].quantile(.99)).hist(bins=50, ax=axes[0])
accounts['pagerank'].clip(upper=accounts['pagerank'].quantile(.99)).hist(bins=50, ax=axes[1])
accounts['community_id'].value_counts().hist(bins=50, ax=axes[2])
axes[0].set_title('In-degree'); axes[1].set_title('PageRank'); axes[2].set_title('Connected component sizes')
plt.tight_layout()
```
